# Overview

Snake game in Pygame with a bot that learns to play by imitating example games. A neural network (MLP) is trained on logged moves, then plays on its own.

# Libraries Used

Pygame: window, drawing, input, game loop

scikit-learn: MLPClassifier (neural network)

NumPy: feature math, loading/saving training data

Python: random, enum, collections.deque, typing

# How it trains

We originally attempted to just train the model on the raw board data, but this resulted in an average score of 0 since we had nowhere near enough data and would probably result in memorization of playing.

To fix this, we condensed the board data down into a few numbers detailed below and trained based on that. This way, the neural network is able to generalize behaviors based on the way its current board position looks rather than just know where to go.

With the old model, if the board is rotated 90 degrees the bot will be completely lost and will likely not have abstracted this information to make a decision on it.

* Features: the board becomes 10 numbers: danger in each direction, apple direction, current heading.

* Data: each move is logged as features + chosen direction to playstyle.txt. A BFS-based expert (gen.py) generated 20,000 rows.

* Training: MLPClassifier(16, 16) learns features to direction.

* Bot: each tick, the MLP predicts the next move. Reverse moves are impossible and are ignored as in the regular game.

Our features that we trained on were essentially looking one step ahead in any direction and detecting if there was "danger" or a game ending variable in the way, as well as one-hot encoding the direction the player is facing and using "direction vectors" which had one for N/S and one for E/W.

I think that we could improve this by using the models results to feed forward into this model, as in predicting death happening moves ahead and updating the vectors accordingly. With this, we could start the model off with no training and then have an algorithm look many moves ahead and determine the optimal move and train on this. This would be like having a professional player play. Also, expand the amount of features we train on since these do not tell much.

# Testing

benchmark.py trains the MLP on playstyle.txt, then plays N games with no window, using the same game and feature code as main.py.

Results:

benchmark.py 100

trained on 229 rows, accuracy 0.865

scores:

100 games | avg 14.92 | median 15 | min 2 | max 33 | std 6.35

Since we ended up getting pretty decent scores on the data copying our playstyles, I would say that this model training works pretty well, especially since just randomly playing would likely end up with a mean score of 0.

# Conclusion

Feeding the raw board to the network failed, because body values shift every tick and the model just memorized. Replacing it with 10 hand-built features fixed this. Putting the chosen move into the inputs also would have made accuracy look perfect while the bot failed in play.

Given more time to train and test, we would expand our model to be self learning, or unassisted, and would hopefully have a higher skill ceiling than the bot that we made.
